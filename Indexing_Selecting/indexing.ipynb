## 🔍 Indexing and Selecting Data

> Indexing in **Pandas** used to **identify, access, and modify subsets** of data in **Series** and **DataFrame** objects.

### Key Purposes of Indexing
- Provides **labels** (metadata) that identify and align data automatically.  
- Enables **intuitive selection** and manipulation of specific rows, columns, or values.


```python
import pandas as pd
import numpy as np
```

## loc
 - Label-based

```python
ser = pd.Series(range(5), index=list("abcde"))
ser
```

```python
ser.loc[["a", "c", "e"]]
```

## iloc

- Index-based

```python
ser.iloc[[0, 2, 4]]
```

## Slicing ranges

```python
s = pd.Series(np.random.randn(10), index=pd.date_range("20230101", periods=10))
s
```

```python
# Get the first five values
s[:5]
```

```python
# Get every second value
s[::2]
```

```python
# Reverse the series
s[::-1]
```

### Selection by callable

```python
df_data = pd.DataFrame(np.random.randn(6, 4), index=list("abcdef"), columns=list("WXYZ"))
df_data
```

```python
df_data.loc[lambda df: df['W'] > 0, :]
```

### Combining positional and label-based indexing

```python
df_combined = pd.DataFrame({"A": [1, 2, 3], "B": [4, 5, 6]})
df_combined
```

```python
#  explicitly getting locations on the indexers, and using positional indexing to select things.
df_combined.loc[df_combined.index[[0, 2]], "A"]
```

```python
# Accessing multiple indexers
df_combined.iloc[0:2, df_combined.columns.get_indexer(["A", "B"])]
```

### Random Sample

```python
s = pd.Series(np.random.randn(10))
s
```

```python
s.sample(5, replace=False)
```

```python
s.sample(5, replace=True)
```

### Setting with enlargement

```python
se = pd.Series(np.random.randn(3))
se
```

```python
se[5] = 10
se
```

```python
dfi = pd.DataFrame(np.random.randn(3, 3), columns=list("ABC"))
dfi
```

```python
dfi.loc[:, 'D'] = dfi.loc[:, 'A'] 
dfi
```
